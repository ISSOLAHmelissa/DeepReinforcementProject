```python
import sys
import os
from collections import defaultdict
from DRL_algorithms.Dynamic_methods import iterative_policy_evaluation_sparse,policy_iteration_sparse,value_iteration_sparse
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))
from Costum_environments.monty_hall_paradox import MontyHallLevel01Env_DP
```

```python
# Exemple de plusieurs politiques différentes
env = MontyHallLevel01Env_DP()

S = env.get_all_states()
A = [0, 1, 2, 'keep', 'switch']
R = [0.0, 1.0]
model = env.get_transition_model()
terminal_states = [('terminal_win',), ('terminal_lose',)]
valid_actions_dict = {s: env.get_valid_actions(s) for s in S}
# Politique aléatoire uniforme
pi_random = {
    s: {a: 1 / len(valid_actions_dict[s]) for a in valid_actions_dict[s]}
    for s in S if s not in terminal_states
}

# Politique toujours "keep" quand possible
pi_keep = {
    s: {a: 1.0 if a == 'keep' else 0.0 for a in valid_actions_dict[s]}
    for s in S if 'keep' in valid_actions_dict[s]
}

# Politique toujours "switch" quand possible
pi_switch = {
    s: {a: 1.0 if a == 'switch' else 0.0 for a in valid_actions_dict[s]}
    for s in S if 'switch' in valid_actions_dict[s]
}

# Évaluation de chaque politique
V_random = iterative_policy_evaluation_sparse(pi_random, S, A, model, terminal_states, valid_actions_dict)
V_keep = iterative_policy_evaluation_sparse(pi_keep, S, A, model, terminal_states, valid_actions_dict)
V_switch = iterative_policy_evaluation_sparse(pi_switch, S, A, model, terminal_states, valid_actions_dict)

# Affichage des valeurs de quelques états clés
print("=== Valeurs estimées sous différentes politiques ===")
etats_extraits = [s for s in S if isinstance(s, tuple) and s[0] == 'second_choice'][:5]
etats_extraits.insert(0, ('start', None, None))  # inclure l’état initial

for s in etats_extraits:
    print(f"État {s} : Random = {V_random[s]:.3f} | Keep = {V_keep.get(s, 0):.3f} | Switch = {V_switch.get(s, 0):.3f}")
```

```python
env = MontyHallLevel01Env_DP()

S = env.get_all_states()
A = [0, 1, 2, 'keep', 'switch']
R = [0.0, 1.0]
model = env.get_transition_model()
terminal_states = [('terminal_win',), ('terminal_lose',)]
valid_actions_dict = {s: env.get_valid_actions(s) for s in S}

pi, V = policy_iteration_sparse(S, A, R, model, terminal_states, valid_actions_dict)

# Affichage complet de la politique et des valeurs V
print("=== Politique optimale (Policy Iteration) ===")
for s in S:
    if s in pi:
        print(f"État {s} → Politique : {pi[s]}")

print("\n=== Valeurs des états V (Policy Iteration) ===")
for s in S:
    print(f"V[{s}] = {V[s]:.3f}")
```
