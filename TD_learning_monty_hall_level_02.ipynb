```python
from Costum_environments.monty_hall_paradox import MontyHallLevel02Env
from DRL_algorithms.td_learning import expected_sarsa_control
```

```python
def print_only_final_choices(policy):
    final_states = {
        state: action
        for state, action in policy.items()
        if state[0] == 'final_choice'
    }
    
    if not final_states:
        print("Aucun état final trouvé.")
        return

    print("=== États finaux : décision switch/keep ===")
    for state, action in sorted(final_states.items(), key=lambda x: x[0][1]):
        last_choice = state[1]
        print(f"Si dernière porte choisie = {last_choice} => Action: {action}")
```

```python
env = MontyHallLevel02Env()
policy_sarsa_monty_02, Q_sarsa_monty_02 = expected_sarsa_control(env,100000)
print(policy_sarsa_monty_02)
print_only_final_choices(policy_sarsa_monty_02)
```
