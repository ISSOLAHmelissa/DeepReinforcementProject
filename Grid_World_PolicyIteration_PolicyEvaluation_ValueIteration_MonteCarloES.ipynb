# Iterative Policy Evaluation

```python
from typing import List
```

```python
def iterative_policy_evaluation(
    pi: np.ndarray,
    S: List[int],
    A: List[int],
    R: List[float],
    T: List[int], # terminal states
    p: np.ndarray,
    theta: float = 0.0001,
    gamma: float = 0.99,
):
  V = np.random.random((len(S),))
  V[T] = 0.0

  while True:
    delta = 0.0

    for s in S:
      v = V[s]
      total = 0.0
      for a in A:
        sub_total = 0.0
        for s_p in S:
          for r_index in range(len(R)):
            r = R[r_index]
            sub_total += p[s, a, s_p, r_index] * (r + gamma * V[s_p])
        total += pi[s, a] * sub_total
      V[s] = total
      abs_diff = np.abs(v - V[s])
      delta = np.maximum(delta, abs_diff)

    if delta < theta:
      break
  return V


```

```python
pi_always_up = np.zeros((len(S), len(A)))
pi_always_up[:, 2] = 1.0  # Toujours Top

V = iterative_policy_evaluation(pi_always_up, S, A, R, T, p)

print("Valeurs de V sous la politique 'always up' :")
print(V)
```

# Policy Iteration

```python
def policy_iteration(
    S: List[int],
    A: List[int],
    R: List[int],
    T: List[int],
    p: np.ndarray,
    theta: float = 0.0001,
    gamma: float = 0.99,
):
  V = np.random.random((len(S),))
  V[T] = 0.0
  pi = np.array([np.random.choice(A) for s in S])
  pi[T] = 0

  while True:

    # Policy Evaluation
    while True:
      delta = 0.0

      for s in S:
        v = V[s]
        total = 0.0
        for s_p in S:
          for r_index in range(len(R)):
            r = R[r_index]
            total += p[s, pi[s], s_p, r_index] * (r + gamma * V[s_p])
        V[s] = total
        abs_diff = np.abs(v - V[s])
        delta = np.maximum(delta, abs_diff)

      if delta < theta:
        break

    # Policy Improvement

    policy_stable = True
    for s in S:
      old_action = pi[s]
      best_a = None
      best_a_score = -999999999.99999
      for a in A:
        score = 0.0
        for s_p in S:
          for r_index in range(len(R)):
            r = R[r_index]
            score += p[s, a, s_p, r_index] * (r + gamma * V[s_p])
        if best_a is None or score > best_a_score:
          best_a = a
          best_a_score = score
      if best_a != old_action:
        policy_stable = False
      pi[s] = best_a

    if policy_stable:
      break

  return pi, V
```

# Value iteration

```python
from typing import List
import numpy as np

def value_iteration(
    S: List[int],
    A: List[int],
    R: List[int],
    T: List[int],
    p: np.ndarray,
    theta: float = 0.0001,
    gamma: float = 0.99,
):
    V = np.zeros(len(S))
    V[T] = 0.0  # Valeur terminale fixée à 0

    while True:
        delta = 0.0
        for s in S:
            if s in T:
                continue
            v = V[s]
            V[s] = max(
                sum(
                    p[s, a, s_p, r_index] * (R[r_index] + gamma * V[s_p])
                    for s_p in S
                    for r_index in range(len(R))
                )
                for a in A
            )
            delta = max(delta, abs(v - V[s]))
        if delta < theta:
            break

    # Extraction de la politique déterministe optimale
    pi = np.zeros(len(S), dtype=int)
    for s in S:
        if s in T:
            continue
        best_a = None
        best_value = -np.inf
        for a in A:
            value = sum(
                p[s, a, s_p, r_index] * (R[r_index] + gamma * V[s_p])
                for s_p in S
                for r_index in range(len(R))
            )
            if value > best_value:
                best_value = value
                best_a = a
        pi[s] = best_a

    return pi, V
```

```python
pi_star, V_star = value_iteration(S, A, R, T, p)
print("Politique optimale :")
print(pi_star)
print("Valeur des états :")
print(V_star)
```

# GridWorld_Monte Carlo

```python
class MonteCarloEnv:

  def num_states(self) -> int:
    raise NotImplementedError()

  def num_actions(self) -> int:
    raise NotImplementedError()

  def step(self, a: int):
    raise NotImplementedError()

  def score(self) -> float:
    raise NotImplementedError()

  def is_game_over(self) -> bool:
    raise NotImplementedError()

  def reset(self):
    raise NotImplementedError()
```

```python
class GridWorldEnv(MonteCarloEnv):
    def __init__(self):
        self.largeur = 5
        self.hauteur = 5
        self.terminal_states = [4, 24]  # T
        self.actions = [0, 1, 2, 3]  # 0: Left, 1: Right, 2: Top, 3: Bottom
        self.reset()

    def num_states(self) -> int:
        return 25  # S = [0, ..., 24]

    def num_actions(self) -> int:
        return len(self.actions)

    def state(self) -> int:
        return self.s

    def step(self, a: int):
        assert a in self.actions
        if self.is_game_over():
            raise Exception("Épisode terminé")

        x, y = self.s % self.largeur, self.s // self.largeur

        # Move agent
        if a == 0 and x > 0:       # left
            x -= 1
        elif a == 1 and x < 4:     # right
            x += 1
        elif a == 2 and y > 0:     # top
            y -= 1
        elif a == 3 and y < 4:     # bottom
            y += 1

        new_state = y * self.largeur + x
        self.s = new_state

        # Apply reward
        if self.s == 4:
            self.inner_score += -3.0
        elif self.s == 24:
            self.inner_score += 1.0
        else:
            self.inner_score += 0.0

    def score(self) -> float:
        return self.inner_score

    def is_game_over(self) -> bool:
        return self.s in self.terminal_states

    def reset(self):
        self.s = 0  # tu peux randomiser ici si nécessaire
        self.inner_score = 0.0
```

```python
import numpy as np
import random
from collections import defaultdict
from typing import List, Tuple

def monte_carlo_es(
    env: MonteCarloEnv,
    gamma: float = 0.99,
    num_episodes: int = 10000,
) -> Tuple[np.ndarray, np.ndarray]:
    S = list(range(env.num_states()))
    A = list(range(env.num_actions()))

    # Initialisation
    pi = np.random.choice(A, size=env.num_states())  # pi[s] = action
    Q = np.zeros((env.num_states(), env.num_actions()))
    Returns = defaultdict(list)  # Returns[(s, a)] = list of returns

    for episode in range(num_episodes):
        # Exploring Starts : on choisit un (s0, a0) aléatoire tel que tout soit possible
        s0 = random.choice(S)
        a0 = random.choice(A)
        env.reset()
        env.s = s0  # Override état initial

        episode_trace = []
        done = False

        # Appliquer la première action "a0" choisie
        env.step(a0)
        r = env.score()
        episode_trace.append((s0, a0, r))
        state = env.state()

        # Continuer selon la politique pi
        while not env.is_game_over():
            action = pi[state]
            env.step(action)
            reward = env.score()
            episode_trace.append((state, action, reward))
            state = env.state()

        # Backward pass
        G = 0.0
        visited = set()

        for t in reversed(range(len(episode_trace))):
            s, a, r = episode_trace[t]
            G = gamma * G + r
            if (s, a) not in visited:
                visited.add((s, a))
                Returns[(s, a)].append(G)
                Q[s, a] = np.mean(Returns[(s, a)])
                pi[s] = np.argmax(Q[s])  # Politique gloutonne

    return Q, pi
```

```python
# Création de l'environnement GridWorld
env = GridWorldEnv()

# Appel de l'algorithme Monte Carlo Exploring Starts
Q, pi = monte_carlo_es(env=env, gamma=0.99, num_episodes=10000)

# Affichage des résultats
print("Politique optimale (pi) :")
print(pi.reshape((5, 5)))  # Pour mieux visualiser sous forme de grille 5x5

print("\nValeurs d'action Q(s, a) :")
print(Q)
```
