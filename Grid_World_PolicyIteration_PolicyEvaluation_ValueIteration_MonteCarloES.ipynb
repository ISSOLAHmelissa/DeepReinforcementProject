# Iterative Policy Evaluation

```python
from typing import List
```

```python
def iterative_policy_evaluation(
    pi: np.ndarray,
    S: List[int],
    A: List[int],
    R: List[float],
    T: List[int], # terminal states
    p: np.ndarray,
    theta: float = 0.0001,
    gamma: float = 0.99,
):
  V = np.random.random((len(S),))
  V[T] = 0.0

  while True:
    delta = 0.0

    for s in S:
      v = V[s]
      total = 0.0
      for a in A:
        sub_total = 0.0
        for s_p in S:
          for r_index in range(len(R)):
            r = R[r_index]
            sub_total += p[s, a, s_p, r_index] * (r + gamma * V[s_p])
        total += pi[s, a] * sub_total
      V[s] = total
      abs_diff = np.abs(v - V[s])
      delta = np.maximum(delta, abs_diff)

    if delta < theta:
      break
  return V


```

```python
pi_always_up = np.zeros((len(S), len(A)))
pi_always_up[:, 2] = 1.0  # Toujours Top

V = iterative_policy_evaluation(pi_always_up, S, A, R, T, p)

print("Valeurs de V sous la politique 'always up' :")
print(V)
```

# Policy Iteration

```python
def policy_iteration(
    S: List[int],
    A: List[int],
    R: List[int],
    T: List[int],
    p: np.ndarray,
    theta: float = 0.0001,
    gamma: float = 0.99,
):
  V = np.random.random((len(S),))
  V[T] = 0.0
  pi = np.array([np.random.choice(A) for s in S])
  pi[T] = 0

  while True:

    # Policy Evaluation
    while True:
      delta = 0.0

      for s in S:
        v = V[s]
        total = 0.0
        for s_p in S:
          for r_index in range(len(R)):
            r = R[r_index]
            total += p[s, pi[s], s_p, r_index] * (r + gamma * V[s_p])
        V[s] = total
        abs_diff = np.abs(v - V[s])
        delta = np.maximum(delta, abs_diff)

      if delta < theta:
        break

    # Policy Improvement

    policy_stable = True
    for s in S:
      old_action = pi[s]
      best_a = None
      best_a_score = -999999999.99999
      for a in A:
        score = 0.0
        for s_p in S:
          for r_index in range(len(R)):
            r = R[r_index]
            score += p[s, a, s_p, r_index] * (r + gamma * V[s_p])
        if best_a is None or score > best_a_score:
          best_a = a
          best_a_score = score
      if best_a != old_action:
        policy_stable = False
      pi[s] = best_a

    if policy_stable:
      break

  return pi, V
```

# Value iteration

```python
from typing import List
import numpy as np

def value_iteration(
    S: List[int],
    A: List[int],
    R: List[int],
    T: List[int],
    p: np.ndarray,
    theta: float = 0.0001,
    gamma: float = 0.99,
):
    V = np.zeros(len(S))
    V[T] = 0.0  # Valeur terminale fixée à 0

    while True:
        delta = 0.0
        for s in S:
            if s in T:
                continue
            v = V[s]
            V[s] = max(
                sum(
                    p[s, a, s_p, r_index] * (R[r_index] + gamma * V[s_p])
                    for s_p in S
                    for r_index in range(len(R))
                )
                for a in A
            )
            delta = max(delta, abs(v - V[s]))
        if delta < theta:
            break

    # Extraction de la politique déterministe optimale
    pi = np.zeros(len(S), dtype=int)
    for s in S:
        if s in T:
            continue
        best_a = None
        best_value = -np.inf
        for a in A:
            value = sum(
                p[s, a, s_p, r_index] * (R[r_index] + gamma * V[s_p])
                for s_p in S
                for r_index in range(len(R))
            )
            if value > best_value:
                best_value = value
                best_a = a
        pi[s] = best_a

    return pi, V
```

```python
pi_star, V_star = value_iteration(S, A, R, T, p)
print("Politique optimale :")
print(pi_star)
print("Valeur des états :")
print(V_star)
```

# GridWorld_Monte Carlo

```python
from tqdm import tqdm
import numpy as np
```

```python
class MonteCarloEnv:

  def num_states(self) -> int:
    raise NotImplementedError()

  def num_actions(self) -> int:
    raise NotImplementedError()

  def step(self, a: int):
    raise NotImplementedError()

  def score(self) -> float:
    raise NotImplementedError()

  def is_game_over(self) -> bool:
    raise NotImplementedError()

  def reset(self):
    raise NotImplementedError()
```

```python
class GridWorld(MonteCarloEnv):
  def __init__(self):
    self.s = 0  # L’agent commence en haut à gauche (état 0) ( position actuel)
    self.inner_score = 0.0

  def num_states(self) -> int:
    return 25  # 5x5 grille

  def num_actions(self) -> int:
    return 4  # 0: gauche, 1: droite, 2: haut, 3: bas

  def state(self) -> int:
    return self.s

  def step(self, a: int):
    if self.is_game_over():
      raise Exception("Épisode terminé")

    row, col = divmod(self.s, 5)  # Convertir état en (ligne, colonne) , pour savoie ou se trouve l'agent pour decider s'il peut bouger ou pas

    # Calculer nouvelle position # la colonne change quand tu vas gauche/droite, la ligne chnage quand tu vas haut/bas
    if a == 0 and col > 0:  # gauche
      col -= 1
    elif a == 1 and col < 4:  # droite
      col += 1
    elif a == 2 and row > 0:  # haut
      row -= 1
    elif a == 3 and row < 4:  # bas
      row += 1

    new_state = row * 5 + col # exemple on est a (2,3) alors new_state = 2*5+3=13, self.s=13
    self.s = new_state

    # Récompense
    if self.s == 4:
      self.inner_score += -3.0
    elif self.s == 24:
      self.inner_score += 1.0
    # sinon reward = 0 (pas besoin de le dire)

  def score(self) -> float:
    return self.inner_score

  def is_game_over(self) -> bool:
    return self.s == 4 or self.s == 24  # États terminaux

  def reset(self):
    self.s = 0  # Retour en haut à gauche
    self.inner_score = 0.0
```

```python
def monte_carlo_exploring_starts(
    env: MonteCarloEnv,
    iterations_count: int,
    gamma: float = 0.9999999,
):
    n_states = env.num_states()
    n_actions = env.num_actions()

    # Initialisation aléatoire de Q et de la politique π (probabilité déterministe ici)
    Q = np.random.random((n_states, n_actions))
    Returns_counts = np.zeros((n_states, n_actions))
    pi = np.zeros((n_states, n_actions))
    for s in range(n_states):
        pi[s, np.argmax(Q[s])] = 1.0  # politique déterministe initiale

    all_states = np.arange(n_states)
    all_actions = np.arange(n_actions)

    for _ in range(iterations_count):
        # Exploring start : état et action de départ aléatoire
        start_state = np.random.choice(all_states)
        start_action = np.random.choice(all_actions)

        env.reset()
        env.s = start_state

        trajectory_states = []
        trajectory_actions = []
        trajectory_rewards = []

        # Première action forcée
        prev_score = env.score()
        env.step(start_action)
        r = env.score() - prev_score

        trajectory_states.append(start_state)
        trajectory_actions.append(start_action)
        trajectory_rewards.append(r)

        # Suite de l’épisode avec la politique pi
        while not env.is_game_over():
            s = env.state()
            a = np.random.choice(all_actions, p=pi[s])
            prev_score = env.score()
            env.step(a)
            r = env.score() - prev_score

            trajectory_states.append(s)
            trajectory_actions.append(a)
            trajectory_rewards.append(r)

        # Calcul des retours G et mise à jour de Q
        G = 0
        for t in reversed(range(len(trajectory_states))):
            s_t = trajectory_states[t]
            a_t = trajectory_actions[t]
            r_t_plus_1 = trajectory_rewards[t]

            G = gamma * G + r_t_plus_1

            # First visit check pour la paire (s,a)
            if (s_t, a_t) not in zip(trajectory_states[:t], trajectory_actions[:t]):
                Returns_counts[s_t, a_t] += 1
                Q[s_t, a_t] += (G - Q[s_t, a_t]) / Returns_counts[s_t, a_t]

                # Mise à jour de la politique : choisir l'action max en Q(s,a)
                best_a = np.argmax(Q[s_t])
                pi[s_t] = 0.0
                pi[s_t, best_a] = 1.0

    return pi, Q
```

```python
# Création de l'environnement GridWorld
env = GridWorld()


# Appel de l'algorithme Monte Carlo Exploring Starts
pi, Q = monte_carlo_exploring_starts(env, iterations_count=10000)


# Affichage des résultats
print("Politique optimale (pi) :")
print(pi.reshape((5, 5)))  # Pour mieux visualiser sous forme de grille 5x5

print("\nValeurs d'action Q(s, a) :")
print(Q)
```
