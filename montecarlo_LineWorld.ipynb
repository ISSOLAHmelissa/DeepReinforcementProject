```python
import sys
import os
import random
from collections import defaultdict
sys.path.append(os.path.abspath(os.path.join(os.getcwd(),"..")))
from secret_envs.monte_carlo_methods import on_policy_first_visit_mc_control,off_policy_mc_control,mc_es_control,generate_episode
from secret_envs.td_learning import expected_sarsa_control,q_learning
from Costum_environments.LineWorld import LineWorldEnv
```

```python
# Exécution de l'entraînement
env = LineWorldEnv()
policy, Q = on_policy_first_visit_mc_control(env, epsilon=0.1, num_episodes=1000)

# Affichage d’un extrait de la politique extraite
print("\n Politique On-Policy extraite (premiers états connus) :")
for i, (state, actions_probs) in enumerate(policy.items()):
    if actions_probs:
        best_action = max(actions_probs.items(), key=lambda x: x[1])[0]
        print(f"État {state} → Action optimale : {LineWorldEnv.INDEX_TO_ACTION[best_action]}")
    if i >= 10:
        break

# Affichage d’un extrait des valeurs Q apprises
print("\n Valeurs Q apprises (extrait) :")
for i, (state, actions) in enumerate(Q.items()):
    q_str = ", ".join(
        f"{LineWorldEnv.INDEX_TO_ACTION[a]}: {v:.3f}" for a, v in actions.items()
    )
    print(f"État {state} → {q_str}")
    if i >= 10:
        break
```

```python
target_policy, Q = off_policy_mc_control(env, num_episodes=1)

# Affichage d’un extrait de la politique extraite
print("\n Politique Off-Policy extraite (premiers états connus) :")
for i, (state, action) in enumerate(target_policy.items()):
    print(f"État {state} → Action optimale : {LineWorldEnv.INDEX_TO_ACTION[action]}")
    if i >= 10:
        break

# Affichage d’un extrait des valeurs Q apprises
print("\n Valeurs Q apprises (extrait) :")
for i, (state, actions_q) in enumerate(Q.items()):
    q_str = ", ".join([f"{LineWorldEnv.INDEX_TO_ACTION[a]}: {v:.3f}" for a, v in actions_q.items()])
    print(f"État {state} → {q_str}")
    if i >= 10:
        break
```

```python
# 1. Création de l'environnement
env = LineWorldEnv()

# 2. Apprentissage avec Monte Carlo Exploring Starts
policy_es, Q_es = mc_es_control(env, num_episodes=1000)

# 3. Affichage d’un extrait de la politique apprise
print(" Politique MC ES extraite (états connus) :")
for i, (state, action_probs) in enumerate(policy_es.items()):
    if isinstance(action_probs, dict):
        best_action = max(action_probs.items(), key=lambda x: x[1])[0]
        best_action_name = LineWorldEnv.INDEX_TO_ACTION[best_action]
    else:
        best_action = action_probs
        best_action_name = LineWorldEnv.INDEX_TO_ACTION[best_action]
    print(f"État {state} → Action optimale : {best_action_name}")
    if i >= 10:
        break

# 4. Affichage d’un extrait des valeurs Q apprises
print("\n Valeurs Q apprises (extrait) :")
for i, (state, actions) in enumerate(Q_es.items()):
    parts = [f"{LineWorldEnv.INDEX_TO_ACTION[a]}: {v:.3f}" for a, v in actions.items()]
    print(f"État {state} → " + ", ".join(parts))
    if i >= 10:
        break
```
