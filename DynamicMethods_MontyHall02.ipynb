```python
import sys
import os
from collections import defaultdict
from DRL_algorithms.Dynamic_methods import iterative_policy_evaluation_sparse,policy_iteration_sparse,value_iteration_sparse
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))
from Costum_environments.monty_hall_paradox import MontyHallLevel02Env_DP
```

```python
env = MontyHallLevel02Env_DP()
S = env.get_all_states()
A_set = set()
for s in S:
    for a in env.get_valid_actions(s):
        A_set.add(a)
A = list(A_set)
R = [0.0, 1.0]
model = env.get_transition_model()
terminal_states = [s for s in S if s[0] == 'terminal']
valid_actions_dict = {s: env.get_valid_actions(s) for s in S}

# Politique uniforme pour tester l’évaluation
pi_uniform = {
    s: {a: 1.0 / len(valid_actions_dict[s]) for a in valid_actions_dict[s]}
    for s in S if s not in terminal_states and len(valid_actions_dict[s]) > 0
}

# Iterative Policy Evaluation Sparse
V_eval = iterative_policy_evaluation_sparse(pi_uniform, S, A, model, terminal_states, valid_actions_dict)

# Affichage
print("\n Valeurs d’évaluation d’une politique uniforme (10 états) :")
for s in list(V_eval.keys())[:10]:
    print(f"État {s} → V(s) = {V_eval[s]:.4f}")
```
