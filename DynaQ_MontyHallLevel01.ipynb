```python
import sys
import os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(),"..")))
from collections import defaultdict
import random
from tqdm import tqdm
from Costum_environments.monty_hall_paradox import MontyHallLevel01Env
from DRL_algorithms.DynaQ import dyna_q_secret
```

```python
env = MontyHallLevel01Env()

Q = dyna_q_secret(env, n_episodes=1000, planning_steps=10)

# Politique extraite
pi_dyna = {
    s: max(a_dict, key=a_dict.get)
    for s, a_dict in Q.items() if a_dict
}
# Affichage de la politique apprise par Dyna-Q
print("Politique Dyna-Q extraite (états connus) :")
for state in pi_dyna:
    print(f"État {state} → Action optimale : {pi_dyna[state]}")

# Affichage des valeurs Q apprises (extrait)
print("\nValeurs Q apprises (extrait) :")
for state in list(Q.keys())[:10]:  # Limite à 10 états pour un affichage partiel
    for action, value in Q[state].items():
        print(f"Q[{state}][{action}] = {value:.3f}")
```

```python

# Simulation
env.reset()
s = env.state_id()
step_count = 0

print("\n Simulation avec politique Dyna-Q :\n")
while not env.is_game_over():
    env.display()
    actions = env.available_actions()
    if s in pi_dyna and pi_dyna[s] in actions:
        a = pi_dyna[s]
    else:
        a = random.choice(actions)
    print(f"Étape {step_count} — état {s} — action choisie : {a}")
    env.step(a)
    s = env.state_id()
    step_count += 1

env.display()
print(f"\n Partie terminée en {step_count} étapes — Score final : {env.score()}")
```
