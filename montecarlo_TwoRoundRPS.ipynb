```python
import sys
sys.path.append("..")  # Va chercher un dossier au-dessus
from collections import defaultdict
import random
from DRL_algorithms.monte_carlo_methods import on_policy_first_visit_mc_control,off_policy_mc_control,mc_es_control
from Costum_environments.TwoRoundRPS import TwoRoundRPS_AgentEnv
```

```python
env = TwoRoundRPS_AgentEnv()
policy, Q = on_policy_first_visit_mc_control(env, epsilon=0.1, num_episodes=1000)

# Affichage formaté
policy_display = {
    state: max(action_probs.items(), key=lambda x: x[1])[0]
    for state, action_probs in policy.items()
}

Q_display = {
    state: {a: round(v, 3) for a, v in actions.items()}
    for state, actions in Q.items()
}
(policy_display, Q_display)
```

```python
target_policy, Q = off_policy_mc_control(env, num_episodes=1000)

print("\nPolitique Off-Policy extraite :")
for state, action in target_policy.items():
    print(f"État {state} → Action optimale : {action}")

print("\nValeurs Q apprises (extrait) :")
for state, action_values in Q.items():
    print(f"État {state} →", end=" ")
    for action, value in action_values.items():
        print(f"{action}: {value:.3f}", end=", ")
    print()
```

```python
env = TwoRoundRPS_AgentEnv()
policy_es, Q_es = mc_es_control(env, num_episodes=1000)

print("Politique MC ES extraite (états connus) :")
for i, (state, action_probs) in enumerate(policy_es.items()):
    if isinstance(action_probs, dict):
        best_action = max(action_probs.items(), key=lambda x: x[1])[0]
    else:
        best_action = action_probs
    print(f"État {state} → Action optimale : {best_action}")
    if i >= 5:
        break

print("\nValeurs Q apprises (extrait) :")
for i, (state, actions_q) in enumerate(Q_es.items()):
    actions_str = ", ".join(f"{a}: {v:.3f}" for a, v in actions_q.items())
    print(f"État {state} → {actions_str}")
    if i >= 5:
        break
```
