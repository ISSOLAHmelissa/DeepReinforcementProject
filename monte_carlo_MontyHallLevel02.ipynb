```python
from Costum_environments.monty_hall_paradox import MontyHallLevel02Env
from DRL_algorithms.monty_carlo_methods import on_policy_fisrt_visit_mc_control
from DRL_algorithms.monty_carlo_methods import off_policy_mc_control
```

```python
def visualize_final_states(policy_data):
    """
    Affiche les politiques des états finaux ('final_choice').

    Args:
        policy_data (dict): Dictionnaire contenant les politiques.

    Exemple d'appel:
        visualize_final_states(your_dictionary)
    """
    print("=== États finaux et leurs politiques ===\n")
    for key, value in policy_data.items():
        if isinstance(key, tuple) and key[0] == 'final_choice':
            state = key[1]
            print(f"État final: {state}")
            if isinstance(value, dict):
                for action, prob in value.items():
                    print(f"  - Action '{action}': probabilité {prob:.3f}")
            else:
                print(f"  - Action choisie: {value}")
            print()
```

```python

env =  MontyHallLevel02Env()
policy_on_mc_monty_02, Q_on_mc_monty_02 = on_policy_fisrt_visit_mc_control(env,0.1,10000)
print(policy_on_mc_monty_02)
visualize_final_states(policy_on_mc_monty_02)
```

```python
policy_off_mc_monty_02, Q_off_mc_monty_02 = off_policy_mc_control(env,10000)
print(policy_off_mc_monty_02)
visualize_final_states(policy_off_mc_monty_02)
```
