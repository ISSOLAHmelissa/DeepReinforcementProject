```python
env =  MontyHallLevel02Env()
agent = HumanAgentMontyHallLevel02(env)
agent.play()
```

```python
def print_only_final_choices(policy):
    print("📊 Politique apprise pour les états 'final_choice' :\n")
    for state, actions in policy.items():
        if state[0] == 'final_choice':
            best_action = max(actions.items(), key=lambda x: x[1])
            print(f"État : {state} ➡️ Meilleure action : {best_action[0]} (proba = {best_action[1]:.2f})")
```

```python

policy_on_mc_monty_02, Q_on_mc_monty_02 = on_policy_fisrt_visit_mc_control(env,0.1,10000)
print(policy_on_mc_monty_02)
print_only_final_choices(policy_on_mc_monty_02)
```

```python
def print_only_final_choices(policy):
    print("🎯 Final Decision Policy (Only 'final_choice' states):\n")
    for state, action in sorted(policy.items(), key=lambda x: str(x[0])):
        if isinstance(state, tuple) and state[0] == 'final_choice':
            last = state[1]
            other = state[2]
            print(f"🔵 Final Choice | Last: {last}, Other: {other} ➡️ Decision: {action.upper()}")
```

```python
policy_off_mc_monty_02, Q_off_mc_monty_02 = off_policy_mc_control(env,10000)
print(policy_off_mc_monty_02)
print_only_final_choices(policy_off_mc_monty_02)
```
