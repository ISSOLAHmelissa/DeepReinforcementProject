```python
import matplotlib.pyplot as plt

def plot_monty_hall_policy(policy):
    """
    Plot a Monty Hall policy as a bar chart.
    
    Parameters:
    - policy (dict): A dictionary with stringified states as keys and actions (int or str) as values.
    """
    # Convert string actions to numeric values for plotting
    action_map = {'keep': 0, 'switch': 1}
    
    x = list(policy.keys())
    y = [action_map[a] if isinstance(a, str) else a for a in policy.values()]
    labels = [str(a) for a in policy.values()]

    plt.figure(figsize=(12, 6))
    bars = plt.bar(x, y, color='skyblue')

    # Add labels on top of bars
    for bar, label in zip(bars, labels):
        plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.05,
                 label, ha='center', va='bottom', fontsize=10)

    plt.xticks(rotation=45, ha='right')
    plt.ylabel('Action')
    plt.yticks([0, 1, 2], ['keep (0)', 'switch (1)', 'door 2 (2)'])  # adjust based on your usage
    plt.title('Monty Hall Policy Visualization')
    plt.tight_layout()
    plt.show()
```

```python
policy_q_learning_monty_01, Q_learning_monty_01 = q_learning(env,500,0.1,0.99,0.9)
print(policy_q_learning_monty_01)
plot_monty_hall_policy(policy_q_learning_monty_01)
```
