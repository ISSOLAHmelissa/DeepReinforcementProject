```python
import sys
import os
import random
sys.path.append(os.path.abspath(os.path.join(os.getcwd(),"..")))
from Costum_environments.grid_world import MonteCarloEnv,GridWorld_MC
from DRL_algorithms.DynaQ import dyna_q_secret
from collections import defaultdict
import random
from tqdm import tqdm
```

```python
# === Apprentissage avec Dyna-Q ===
env = GridWorld_MC()
Q = dyna_q_secret(env, n_episodes=300, planning_steps=20)

# === Extraction de la politique ===
pi_dyna = {
    s: max(a_dict, key=a_dict.get)
    for s, a_dict in Q.items() if a_dict
}

# === Affichage de la politique extraite (états connus) ===
print("Politique Dyna-Q extraite (extrait) :")
for s in sorted(pi_dyna.keys()):
    print(f"État {s} → Action optimale : {pi_dyna[s]}")

# === Affichage des valeurs Q apprises (extrait) ===
print("\nValeurs Q apprises (extrait) :")
for s in sorted(Q.keys()):
    for a, q_val in Q[s].items():
        print(f"Q[{s}][{a}] = {q_val:.3f}")
```

```python
# Simulation de la politique apprise
env.reset()
s = env.state_id()
step_count = 0

print("\n Simulation avec politique Dyna-Q :\n")
while not env.is_game_over():
    env.display()
    actions = env.available_actions()
    if s in pi_dyna and pi_dyna[s] in actions:
        a = pi_dyna[s]
    else:
        a = random.choice(actions)
    print(f"Étape {step_count} — État {s} — Action {a}")
    env.step(a)
    s = env.state_id()
    step_count += 1

env.display()
print(f"\n Partie terminée en {step_count} étapes — Score : {env.score()}")
```
