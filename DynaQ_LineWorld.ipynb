```python
import sys
import os
import random
sys.path.append(os.path.abspath(os.path.join(os.getcwd(),"..")))
from collections import defaultdict
import random
from tqdm import tqdm
from Costum_environments.LineWorld import LineWorldEnv
from DRL_algorithms.DynaQ import dyna_q_secret
```

```python
# 1. Création de l’environnement
env = LineWorldEnv()

# 2. Entraînement avec Dyna-Q
Q = dyna_q_secret(env, n_episodes=200, planning_steps=10)

# 3. Extraction de la politique
pi_dyna = {
    s: max(a_dict, key=a_dict.get)
    for s, a_dict in Q.items() if a_dict
}

# 4. Affichage des résultats
print("\n Politique Dyna-Q extraite (états connus) :")
for state, action in pi_dyna.items():
    print(f"État {state} → Action optimale : {LineWorldEnv.INDEX_TO_ACTION[action]}")

print("\n Valeurs Q apprises (extrait) :")
for state, actions in Q.items():
    readable_actions = {LineWorldEnv.INDEX_TO_ACTION[a]: round(v, 3) for a, v in actions.items()}
    print(f"État {state} → {readable_actions}")
```

```python

# Simulation
env.reset()
s = env.state_id()
step_count = 0

print("\n Simulation avec politique Dyna-Q :\n")
while not env.is_game_over():
    env.display()
    actions = env.available_actions()
    if s in pi_dyna and pi_dyna[s] in actions:
        a = pi_dyna[s]
    else:
        a = random.choice(actions)
    print(f"Étape {step_count} — État {s} — Action {a}")
    env.step(a)
    s = env.state_id()
    step_count += 1

env.display()
print(f"\n Partie terminée en {step_count} étapes — Score : {env.score()}")
```
