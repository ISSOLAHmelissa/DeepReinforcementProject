```python
import sys
import os
import random
sys.path.append(os.path.abspath(os.path.join(os.getcwd(),"..")))
from DRL_algorithms.DynaQ import dyna_q_secret
from secret_envs_wrapper import SecretEnv3
```

```python
env = SecretEnv3()
Q = dyna_q_secret(env, n_episodes=300, planning_steps=15)

# Extraire la politique :
pi_dyna = {
    s: max(a_dict, key=a_dict.get)
    for s, a_dict in Q.items()
    if a_dict  # ✅ on garde uniquement les dictionnaires non vides
}

# Affichage de la politique extraite
print("\nPolitique Dyna-Q extraite (états connus) :")
for i, (state, action) in enumerate(pi_dyna.items()):
    print(f"État {state} → Action optimale : {action}")
    if i >= 9:
        break

# Affichage des valeurs Q associées
print("\nValeurs Q associées (états connus) :")
for i, (state, actions) in enumerate(Q.items()):
    print(f"État {state} → {actions}")
    if i >= 9:
        break
```

```python
env.reset()
s = env.state_id()
step_count = 0

print("▶️ Simulation avec politique Dyna-Q :\n")
while not env.is_game_over():
    env.display()
    actions = env.available_actions()
    if s in pi_dyna and pi_dyna[s] in actions:
        a = pi_dyna[s]
    else:
        a = random.choice(actions)
    print(f"Étape {step_count} — état {s} — action choisie : {a}")
    env.step(a)
    s = env.state_id()
    step_count += 1

env.display()
print(f"\n✅ Partie terminée en {step_count} étapes — Score final : {env.score()}")
```
