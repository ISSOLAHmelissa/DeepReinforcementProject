```python
from secret_envs_wrapper import SecretEnv3
from collections import defaultdict
import sys
import os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))
from secret_envs.td_learning import q_learning_new,sarsa
```

```python
# 1. Initialiser l’environnement
env = SecretEnv3()

# 2. Exécuter Q-Learning
policy_q, Q_table = q_learning_new(env, num_episodes=1000, alpha=0.1, gamma=0.99, epsilon=0.1)

# 3. Afficher un extrait de la politique apprise
print("\nPolitique Q-Learning extraite (états connus) :")
for i, (state, action) in enumerate(policy_q.items()):
    print(f"État {state} → Action optimale : {action}")
    if i >= 9:
        break

# 4. Afficher les Q-valeurs associées pour les mêmes états
print("\nValeurs Q associées (états connus) :")
for i, (state, q_values) in enumerate(Q_table.items()):
    print(f"État {state} → {q_values}")
    if i >= 9:
        break
```

```python
# 1. Initialiser l’environnement
env = SecretEnv3()

# 2. Apprentissage via Expected SARSA
policy, Q = sarsa(
    env=env,
    num_episodes=1000,
    alpha=0.1,
    gamma=0.99,
    epsilon=0.1
)

# 3. Afficher un extrait de la politique
print("\n Politique Expected SARSA extraite (états connus) :")
for i, (state, action) in enumerate(policy.items()):
    print(f"État {state} → Action optimale : {action}")
    if i >= 9:
        break

# 4. Afficher un extrait des Q-valeurs
print("\n Valeurs Q associées (états connus) :")
for i, (state, q_values) in enumerate(Q.items()):
    print(f"État {state} → {q_values}")
    if i >= 9:
        break
```
