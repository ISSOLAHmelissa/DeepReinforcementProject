EXPLORING THE SECRET ENV 2

```python
env = SecretEnv2()

print("Nombre d'états:", env.num_states())
print("Nombre d'actions:", env.num_actions())
print("Nombre de récompenses:", env.num_rewards())

for i in range(env.num_rewards()):
    print(f"Récompense[{i}] =", env.reward(i))
```

```python
env.reset()
initial_state = env.state_id()
print("État initial :", initial_state)

actions = env.available_actions()
print("Actions disponibles depuis l'état initial :", actions)
```

```python
secret_env_2_mc_on_policy , secret_env_2_mc_on_Q = on_policy_first_visit_mc_control(env,0.1,10)
print(secret_env_2_mc_on_policy)
```

```python
secret_env_2_mc_off_policy , secret_env_2_mc_off_Q = off_policy_mc_control(env,10)
print(secret_env_2_mc_off_policy)
```

```python
secret_env_2_sarsa_policy , secret_env_2_sarsa_Q = expected_sarsa_control(env,1000)
print(secret_env_2_mc_off_policy)
```
