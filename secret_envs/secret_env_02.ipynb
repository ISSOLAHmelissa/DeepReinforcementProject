```python
def run_episode_with_policy(env: SecretEnv2, policy: dict, render: bool = True):
    env.reset()
    state = env.state_id()
    total_reward = 0
    prev_score = env.score()
    steps = 0

    while not env.is_game_over():
        if render:
            env.display()
        
        # Vérifie que l'état est connu dans la politique
        if state not in policy:
            print(f"État {state} non trouvé dans la politique. Arrêt.")
            break

        action = int(policy[state])

        # Vérifie que l'action est autorisée
        if env.is_forbidden(action):
            print(f"Action {action} interdite dans l'état {state}. Arrêt.")
            break

        # Appliquer l'action
        env.step(action)
        new_score = env.score()
        reward = new_score - prev_score
        prev_score = new_score
        total_reward += reward

        # Passage à l'état suivant
        state = env.state_id()
        steps += 1

    print(f"Épisode terminé en {steps} étapes avec un score total de {total_reward}")
```

```python

run_episode_with_policy(env, secret_env_2_sarsa_policy)
```
