```python
import sys
import os
from collections import defaultdict
import random
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))
from Costum_environments.TwoRoundRPS import TwoRoundRPS_AgentEnv
from DRL_algorithms.td_learning import expected_sarsa_control,q_learning_new
```

```python
env = TwoRoundRPS_AgentEnv()

# Apprentissage Q-Learning
policy_q, Q_q = q_learning_new(env, num_episodes=1000, alpha=0.1, gamma=0.95, epsilon=0.1)

# Affichage de la politique apprise (états connus)
print("Politique Q-Learning extraite (états connus) :")
for state in list(policy_q.keys())[:10]:
    print(f"État {state} → Action optimale : {policy_q[state]}")

# Affichage des valeurs Q apprises
print("\nValeurs Q apprises (extrait) :")
for state in list(Q_q.keys())[:10]:
    q_values = Q_q[state]
    for action, value in q_values.items():
        print(f"État {state} → {action}: {value:.3f}")
```

```python
env = TwoRoundRPS_AgentEnv()

# 2. Apprentissage avec Expected SARSA
policy, Q = expected_sarsa_control(
    env=env,
    num_episodes=1000,
    alpha=0.1,
    gamma=0.99,
    epsilon=0.1
)

# 3. Affichage d’un extrait de la politique
print("Politique Expected SARSA extraite (états connus) :")
for i, (state, action) in enumerate(policy.items()):
    print(f"État {state} → Action optimale : {action}")
    if i >= 9:
        break
# 3. Affichage des valeurs Q apprises
print("\nValeurs Q apprises (extrait) :")
for i, (state, actions_q) in enumerate(Q.items()):
    line = f"État {state} → "
    line += ", ".join(f"{a}: {round(q, 3)}" for a, q in actions_q.items())
    print(line)
    if i >= 9:
        break
```
