```python

env = LineWorldEnv()

# 2. Exécuter Q-Learning
policy_q, Q_table = q_learning(env, num_episodes=1000, alpha=0.1, gamma=0.99, epsilon=0.1)

# 3. Affichage de la politique
print("\n Politique Q-Learning extraite (états connus) :")
for i, (state, action) in enumerate(policy_q.items()):
    action_name = LineWorldEnv.INDEX_TO_ACTION[action]
    print(f"État {state} → Action optimale : {action_name}")
    if i >= 10:
        break

print(Q_table)
```
