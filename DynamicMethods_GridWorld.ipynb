```python
import sys
import os
from collections import defaultdict
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))
from Costum_environments.grid_world import GridWorld
from DRL_algorithms.Dynamic_methods import iterative_policy_evaluation_sparse,policy_iteration_sparse,value_iteration_sparse
```

```python
# === Initialisation de l'environnement GridWorld ===
env = GridWorld()

S = env.get_all_states()
A = env.get_all_actions()
R = env.get_rewards()

# Conversion du modèle
model = defaultdict(lambda: defaultdict(list))
for s in S:
    for a in A:
        for (p, s_p, r, done) in env.get_model()[s][a]:
            model[s][a].append((p, s_p, r, done))



valid_actions_dict = env.get_valid_actions()
terminal_states = env.get_terminal_states()
```

```python
# === Politique aléatoire uniforme ===
pi_random = {
    s: {a: 1 / len(valid_actions_dict[s]) for a in valid_actions_dict[s]}
    for s in S if s not in terminal_states
}

# === Politique qui favorise toujours l'action "down" si disponible ===
pi_down = {
    s: {a: 1.0 if a == 'down' else 0.0 for a in valid_actions_dict[s]}
    for s in S if s not in terminal_states
}

# === Politique qui favorise toujours l'action "right" si disponible ===
pi_right = {
    s: {a: 1.0 if a == 'right' else 0.0 for a in valid_actions_dict[s]}
    for s in S if s not in terminal_states
}

# === Évaluation des politiques ===
V_random = iterative_policy_evaluation_sparse(pi_random, S, A, model, terminal_states, valid_actions_dict)
V_down = iterative_policy_evaluation_sparse(pi_down, S, A, model, terminal_states, valid_actions_dict)
V_right = iterative_policy_evaluation_sparse(pi_right, S, A, model, terminal_states, valid_actions_dict)

# === Affichage des résultats pour quelques états choisis ===
print("=== Valeurs estimées sous 3 politiques différentes ===")
for s in [0, 1, 2, 3, 4, 20, 21, 22, 23, 24]:
    print(f"État {s} : Aléatoire = {V_random[s]:.3f} | Down = {V_down[s]:.3f} | Right = {V_right[s]:.3f}")
```

```python
# === Politique et valeurs optimales via Policy Iteration ===
pi_pi, V_pi = policy_iteration_sparse(
    S=S,
    A=A,
    R=R,
    model=model,
    terminal_states=terminal_states,
    valid_actions_dict=valid_actions_dict
)

# Affichage de la politique optimale (extrait)
print("Politique optimale (extrait) :")
for s in [0, 1, 2, 3, 20, 21, 22, 23]:
    if s in pi_pi:
        best_action = max(pi_pi[s], key=pi_pi[s].get)
        print(f"État {s} → action optimale : {best_action}")

# Affichage des valeurs optimales (extrait)
print("\nValeurs de V_pi (extrait) :")
for s in [0, 1, 2, 3, 20, 21, 22, 23]:
    print(f"V[{s}] = {V_pi[s]:.3f}")
```

```python

pi_opt, V_opt = value_iteration_sparse(
    S=S,
    A=A,
    R=R,
    model=model,
    terminal_states=terminal_states,
    valid_actions_dict=valid_actions_dict
)

# === Affichage de la politique optimale et des valeurs optimales ===
print("Politique optimale (complète) :")
for s in S:
    if s not in terminal_states and s in pi_opt:
        action_opt = max(pi_opt[s], key=pi_opt[s].get)
        print(f"État {s} → action optimale : {action_opt}")

print("\nValeurs de V_opt (complètes) :")
for s in S:
    print(f"V_opt[{s}] = {V_opt[s]:.3f}")
```
